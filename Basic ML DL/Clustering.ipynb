```python


# 1. Define Paths (Re-defining BASE_DIR to ensure no NameError)
BASE_DIR = '/content/drive/MyDrive/Colab Notebooks/Data'

train_pos_path = os.path.join(BASE_DIR, 'train', 'pos')
train_neg_path = os.path.join(BASE_DIR, 'train', 'neg')
train_unsup_path = os.path.join(BASE_DIR, 'train', 'unsup')

# 2. Loading Function
def load_reviews(folder, sentiment):
    reviews = []
    if not os.path.exists(folder):
        print(f"⚠️ Warning: Folder {folder} not found.")
        return pd.DataFrame()
    for filename in os.listdir(folder):
        if filename.endswith(".txt"):
            with open(os.path.join(folder, filename), "r", encoding="utf-8") as f:
                reviews.append(f.read())
    return pd.DataFrame({"review": reviews, "sentiment": sentiment})

# 3. Load all 3 parts of the training data
df_pos = load_reviews(train_pos_path, "positive")
df_neg = load_reviews(train_neg_path, "negative")
df_unsup = load_reviews(train_unsup_path, "unlabeled")

# Combine them
df_all = pd.concat([df_pos, df_neg, df_unsup]).reset_index(drop=True)

print(f"✅ Data Loaded!")
print(f"Total reviews: {len(df_all)}")
print(f"Breakdown:\n{df_all['sentiment'].value_counts()}")
```
